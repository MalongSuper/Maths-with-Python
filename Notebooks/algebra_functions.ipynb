## Algebraic Functions as Graphs

Algebraic functions can be represented graphically, allowing us to visualize their behavior and properties. The graph of an algebraic function is a set of points in the coordinate plane that satisfy the function's equation.

```python
import numpy as np
import scipy as sp  # Might be useful for more advanced mathematical functions
import matplotlib.pyplot as plt
```

```python
# Define a function that takes a function as parameter and then plot it
def plot_function(func, x_range=(-10, 10)):
    
    # Convert to lambda
    if isinstance(func, str):
        lambda_func = eval(f"lambda x: {func}")

    x = np.linspace(x_range[0], x_range[1], 100)
    y = lambda_func(x)

    plt.plot(x, y)
    plt.title(f'Plot of {func}')
    plt.xlabel('x')
    plt.ylabel('f(x)')
    plt.grid()
    plt.show()
```

#### 1. Linear Function

A linear function is of the form $f(x) = mx + b$, where $m$ is the slope and $b$ is the y-intercept. The graph of a linear function is a straight line.

```python
# Plot a linear function
f = "2 * x + 1"
plot_function(f)
```

#### 2. Quadratic Function

A quadratic function is of the form $f(x) = ax^2 + bx + c$, where $a$, $b$, and $c$ are constants and $a \neq 0$. The graph of a quadratic function is a parabola, which can open upwards or downwards depending on the sign of $a$.

```python
# Plot a quadratic function
f = "x**2 - 4*x + 3"
plot_function(f)
```

#### 3. Cubic Function

A cubic function is of the form $f(x) = ax^3 + bx^2 + cx + d$, where $a$, $b$, $c$, and $d$ are constants and $a \neq 0$. The graph of a cubic function can have one or two turning points and can exhibit various shapes depending on the coefficients.

```python
# Plot a cubic function
f = "x**3 - 2*x**2 - 5*x + 6"
plot_function(f)
```

#### 4. Exponential Function

An exponential function is of the form $f(x) = a \cdot b^x$, where $a$ is a constant, $b$ is the base (with $b > 0$ and $b \neq 1$), and $x$ is the exponent. The graph of an exponential function can either grow rapidly or decay, depending on the value of the base $b$.

```python
# Plot an exponential function
f = "2**x"
plot_function(f)
```

#### 5. Power Function

A general power function is of the form $f(x) = ax^n$, where $a$ is a constant and $n$ is a real number. The graph of a power function can take various shapes depending on the value of $n$. 

For example, if $n$ is a positive integer, the graph will have a polynomial shape, while if $n$ is negative, the graph will have a hyperbolic shape.

```python
# Plot a power function
f = "x**(1/2) - 4*x + 3"
plot_function(f)
```

#### 6. Absolute Value Function

An absolute value function is of the form $f(x) = a|x - h| + k$, where $a$, $h$, and $k$ are constants. The graph of an absolute value function is a V-shaped graph that opens upwards or downwards depending on the sign of $a$.

```python
# Plot an absolute value function
f = "abs(x - 2) + 1"
plot_function(f)
```

#### 7. Logarithmic Function

Logarithmic functions are the inverse of exponential functions and are of the form $f(x) = a \cdot \log_b(x - h) + k$, where $a$, $b$, $h$, and $k$ are constants, and $b > 0$ and $b \neq 1$. The graph of a logarithmic function increases slowly and has a vertical asymptote at $x = h$.

```python
# Plot a logarithmic function
f = "np.log2(x)" 
plot_function(f)
```

#### 8. Square Root Function

A square root function is of the form $f(x) = a\sqrt{x - h} + k$, where $a$, $h$, and $k$ are constants. The graph of a square root function is a curve that starts at the point $(h, k)$ and increases gradually as $x$ increases.

```python
# Plot a square root function
f = "np.sqrt(x - 1) + 2"
plot_function(f)
```

#### 9. Cube Root Function

Different from the square root function, a cube root function is of the form $f(x) = a\sqrt[3]{x - h} + k$, where $a$, $h$, and $k$ are constants. The graph of a cube root function has an S-shaped curve that passes through the point $(h, k)$ and extends infinitely in both directions.

```python
# Plot a cubic root function
f = "np.cbrt(x)"
plot_function(f)
```

#### 10. Reciprocal Function

Reciprocal functions are of the form $f(x) = \frac{a}{x - h} + k$, where $a$, $h$, and $k$ are constants. The graph of a reciprocal function has two branches, one in the first quadrant and one in the third quadrant, and has vertical and horizontal asymptotes at $x = h$ and $y = k$, respectively.

```python
# Plot a reciprocal function
f = "1 / x"
plot_function(f)
```

#### 11. Trigonometric Functions

Trigonometric functions, such as sine, cosine, and tangent, are periodic functions that can be represented graphically. The graphs of these functions exhibit wave-like patterns and have specific properties, such as amplitude, period, and phase shift.

- Sine Function: $f(x) = a \cdot \sin(bx + c) + d$

- Cosine Function: $f(x) = a \cdot \cos(bx + c) + d$

- Tangent Function: $f(x) = a \cdot \tan(bx + c) + d$

- Cotangent Function: $f(x) = a \cdot \cot(bx + c) + d$

```python
# Here we will draw all four types of functions in four plots 
# but in one figure for comparison
x = np.linspace(-10, 10, 100)
sin_x = np.sin(x)
cos_x = np.cos(x)
tan_x = np.tan(x)
cot_x = 1 / np.tan(x)

# Create a figure with 2x2 subplots
fig, axs = plt.subplots(2, 2, figsize=(10, 8))

axs[0, 0].plot(x, sin_x)
axs[0, 0].set_title('Sine Function')
axs[0, 0].grid()

axs[0, 1].plot(x, cos_x)
axs[0, 1].set_title('Cosine Function')
axs[0, 1].grid()

axs[1, 0].plot(x, tan_x)
axs[1, 0].set_title('Tangent Function')
axs[1, 0].set_ylim(-10, 10)  # Limit y-axis for better visualization
axs[1, 0].grid()

axs[1, 1].plot(x, cot_x)
axs[1, 1].set_title('Cotangent Function')
axs[1, 1].set_ylim(-10, 10)  # Limit y-axis for better visualization
axs[1, 1].grid()

plt.suptitle('Trigonometric Functions')
plt.tight_layout()
plt.show()
```

#### 12. Calculus Functions

Calculus functions, such as derivatives and integrals, can also be represented graphically. The graph of a derivative function represents the slope of the original function at each point, while the graph of an integral function represents the area under the curve of the original function.

##### 12.1. Derivative to Detect Local and Global Maxima and Minima

The derivative of a function can be used to find local and global maxima and minima. By analyzing the critical points where the derivative is zero or undefined, we can determine the behavior of the function and identify points of interest.

Let $f(x)$ be a differentiable function. The critical points are found by solving the equation: $f'(x) = 0$

```python
def plot_derivative(func, x_range=(-10, 10)):
    # Convert to lambda
    if isinstance(func, str):
        lambda_func = eval(f"lambda x: {func}")

    # Create an array of x values
    x = np.linspace(x_range[0], x_range[1], 100)
    # Calculate the derivative using numpy's gradient function
    y = lambda_func(x)
    dy_dx = np.gradient(y, x)  # Numerical derivative

    # Determine the local maxima and minima
    local_maxima = (np.diff(np.sign(dy_dx)) < 0).nonzero()[0] + 1
    local_minima = (np.diff(np.sign(dy_dx)) > 0).nonzero()[0] + 1
    print(f"Local maxima at x = {x[local_maxima]} with f(x) = {y[local_maxima]}")
    print(f"Local minima at x = {x[local_minima]} with f(x) = {y[local_minima]}")

    # Plot the original function and its derivative
    plt.figure(figsize=(10, 6))
    plt.plot(x, y, label=f'f(x) = {func}', color='blue')
    plt.plot(x, dy_dx, label=f"f'(x) = derivative of {func}", color='red')
    plt.title(f'Function and its Derivative')
    plt.xlabel('x')
    plt.ylabel('f(x) and f\'(x)')
    plt.axhline(0, color='black', lw=0.5, ls='--')
    plt.axvline(0, color='black', lw=0.5, ls='--')
    plt.legend()
    plt.grid()
    plt.show()


# Plot the derivative of a quadratic function
f = "x**3 - 3*x"
plot_derivative(f)
```

##### 12.2. Integral to Calculate Area Under the Curve

Integration can be used to calculate the area under the curve of a function. The definite integral of a function over a specific interval gives the total area between the graph of the function and the x-axis within that interval.

Formula:

$$A = \int_{a}^{b} f(x) dx$$

Where $f(x)$ is the function, and $[a, b]$ is the interval over which we want to calculate the area.

```python
def plot_area_under_curve(func, x_range=(-10, 10)):
    # Convert to lambda
    if isinstance(func, str):
        lambda_func = eval(f"lambda x: {func}")

    # Create an array of x values
    x = np.linspace(x_range[0], x_range[1], 100)
    y = lambda_func(x)

    # Calculate the area under the curve using numerical integration
    area = np.trapezoid(y, x)
    print(f"Area under the curve: Approximately {area}")

    # Plot the function and shade the area under the curve
    plt.figure(figsize=(10, 6))
    plt.plot(x, y, label=f'f(x) = {func}', color='blue')
    plt.fill_between(x, y, where=(y > 0), color='lightblue', alpha=0.5, label='Area under curve')
    plt.title(f'Area Under the Curve of {func}')
    plt.xlabel('x')
    plt.ylabel('f(x)')
    plt.axhline(0, color='black', lw=0.5, ls='--')
    plt.axvline(0, color='black', lw=0.5, ls='--')
    plt.legend()
    plt.grid()
    plt.show()

# Example 1: Plot the area under the curve of a simple quadratic function
f = "x**2"
plot_area_under_curve(f, x_range=(-5, 5))
```

```python
# Example 2: Plot the area under the curve of a cubic function
f = "x**3 - 4*x**2 + 3*x"
plot_area_under_curve(f)
```

##### 12.2. Integral to Calculate Area Between the Curve

Also, integration can be used to calculate the area between two curves. By finding the definite integral of the difference between the two functions over a specific interval, we can determine the total area enclosed by the curves.

Formula:

$$A = \int_{a}^{b} |f(x) - g(x)| dx$$

Where $f(x)$ and $g(x)$ are the two functions, and $[a, b]$ is the interval over which we want to calculate the area.

```python
# Two functions and their area between the curve
def plot_area_between_curves(func1, func2, x_range=(-10, 10)):
    # Convert to lambda
    if isinstance(func1, str):
        lambda_func1 = eval(f"lambda x: {func1}")
    if isinstance(func2, str):
        lambda_func2 = eval(f"lambda x: {func2}")   

    # Create an array of x values
    x = np.linspace(x_range[0], x_range[1], 100)
    y1 = lambda_func1(x)
    y2 = lambda_func2(x)

    # Calculate the area between the curves using numerical integration
    area = np.trapezoid(np.abs(y1 - y2), x)
    print(f"Area between the curves: Approximately {area}")

    # Plot the functions and shade the area between the curves
    plt.figure(figsize=(10, 6))
    plt.plot(x, y1, label=f'f1(x) = {func1}', color='blue')
    plt.plot(x, y2, label=f'f2(x) = {func2}', color='red')
    plt.fill_between(x, y1, y2, where=(y1 > y2), color='lightblue', alpha=0.5, label='Area between curves')
    plt.fill_between(x, y1, y2, where=(y1 < y2), color='lightcoral', alpha=0.5)
    plt.title(f'Area Between the Curves of {func1} and {func2}')
    plt.xlabel('x')
    plt.ylabel('f(x)')
    plt.axhline(0, color='black', lw=0.5, ls='--')
    plt.axvline(0, color='black', lw=0.5, ls='--')
    plt.legend()
    plt.grid()
    plt.show()

# Plot the area between two functions
f1 = "x**2"
f2 = "x"
plot_area_between_curves(f1, f2)
```

The second example demonstrates an intersting case when the curve changes its position within the interval. In such case, the area is calculated by splitting the integral into two parts, one for each curve, and then summing the absolute values of the areas.

- For example, for $x^3 - 4 \times x ^ 2 + 3$ and $x^2 + 3 \times x - 1$. Find their intersection points and calculate the area between them: 

- $x^3 - 4 \times x ^ 2 + 3 = x^2 + 3 \times x - 1$ results in $x^3 - 5 \times x ^ 2 - 3 \times x + 4 = 0$ which has three roots: $x_1 = -1$, $x_2 = 1$, and $x_3 = 4$. The area between the curves can be calculated as:

$$A = \int_{-1}^{1} |(x^3 - 4 \times x ^ 2 + 3) - (x^2 + 3 \times x - 1)| dx + \int_{1}^{4} |(x^3 - 4 \times x ^ 2 + 3) - (x^2 + 3 \times x - 1)| dx$$

```python
# Plot the area between two functions
f1 = "x**3 - 4 * x ** 2 + 3"
f2 = "x**2 + 3 * x - 1"
plot_area_between_curves(f1, f2)
```
